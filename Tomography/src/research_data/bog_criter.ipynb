Рассмотрим критерий основанный на матрице информации H, для которой протокол считается полным, если $r^2$ собственных значений равно 0, а (2s-r)r строго > 0. Для начала рассмотрим протокол приведённый в статье "Statistical reconstruction of qutrits"(2004), https://sci-hub.ru/https://doi.org/10.1103/PhysRevA.70.042303 .

```python
from purification_state import*
rank = [1,2,3]
dim = 3
for r in rank:
    pure_class = Purification(rank = r, dimension = dim)
    val = (np.real(np.round(pure_class.value_matrix_information(), 14)))
    val = list(np.sign(val))
    numb_zero = val.count(0)
    numb_positiv = val.count(1)
    
    print(f"Количество теоретических нулей = {r**2}, количество нулей для протокола = {numb_zero}")
    print(f"Количество теоретических положительных чисел = {(2*dim-r)*r}, количество положительных чисел для протокола = {numb_positiv}")
```

Теперь посчитаем для наших протоколов. Для протокола с пластинкой $\lambda/4$.

```python
rank = [1,2,3]
dim = 3
for r in rank:
    start_protocol = [Gl_4(0),Gl_4(np.pi/8),Gl_4(3*np.pi/8),Gl_4(5*np.pi/8),Gl_4(7*np.pi/8)]
    pure_class = Purification(rank = r, dimension = dim, protocol = start_protocol)
    val = (np.real(np.round(pure_class.value_matrix_information(), 15)))
    print(np.round(sorted(val),16))
    val = list(np.sign(val))
    numb_zero = val.count(0)
    numb_positiv = val.count(1)
    print(f"Количество теоретических нулей = {r**2}, количество нулей для протокола = {numb_zero}")
```

Проверка всех переборов элементов протокола для выбора 3 элементов 

```python
import itertools

rank = [1,2,3]
dim = 3
all_prot = [np.diag((1,1,1)), Gl_2(np.pi/8), Gl_8(0) @ Gl_2(np.pi/8), Gl_4(0), Gl_4(np.pi/8), Gl_4(3*np.pi/8), Gl_4(5*np.pi/8), Gl_4(7*np.pi/8)]

for triple in itertools.combinations(list(range(len(all_prot))), 3):
    ind_1, ind_2, ind_3 = triple
    print(f"\n\n Индексы элементов протокола: {ind_1, ind_2, ind_3}")
    for r in rank:
        start_protocol = [all_prot[ind_1], all_prot[ind_2], all_prot[ind_3]]
        pure_class = Purification(rank = r, dimension = dim, protocol = start_protocol)
        val = (np.real(np.round(pure_class.value_matrix_information(), 14)))
        # print(np.round(sorted(val),16))
        val = list(np.sign(val))
        numb_zero = val.count(0)
        numb_positiv = val.count(1)
        print(f"Количество теоретических нулей = {r**2}, количество нулей для протокола = {numb_zero}")
```
